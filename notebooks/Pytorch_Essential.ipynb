# Simple Neural Network

## Import NN Module and Create Dummy Data

```python
import torch.nn as nn
## our data in tensor form
x = torch.tensor([[-1.0],  [0.0], [1.0], [2.0], [3.0], [4.0]], dtype=torch.float)
y = torch.tensor([[-3.0], [-1.0], [1.0], [3.0], [5.0], [7.0]], dtype=torch.float)
```

$$y=2x-1$$

```python
import matplotlib.pyplot as plt
plt.plot(x.numpy(), y.numpy())
plt.scatter(x.numpy(), y.numpy())
plt.show()
```

## Define Neural Network

```python
## Neural network with 1 hidden layer
layer1 = nn.Linear(1,1, bias=False)
model = nn.Sequential(layer1)
```

## Loss and Optimizer

```python
## loss function
criterion = nn.MSELoss()

## optimizer algorithm
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

## Training the Neural Network Model

```python
## training
epoch = 150
for i in range(150):
    model = model.train()
    train_running_loss = 0.0

    ## forward
    output = model(x)
    loss = criterion(output, y)
    optimizer.zero_grad()

    ## backward + update model params
    loss.backward()
    optimizer.step()
    train_running_loss += loss.detach().item()

    model.eval()
    print('Epoch: %d | Loss: %.4f' %(i, train_running_loss) )
```

## Evaliating the Model

```python
## test the model
sample = torch.tensor([5.0], dtype=torch.float)
predicted = model(sample)
print(predicted.detach().item())
```
