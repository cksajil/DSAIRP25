# Linear Regression Algorithm

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
np.random.seed(42) # for reproducibility
n_samples = 100

# Generate two features (X1, X2)
X1 = 2 * np.random.rand(n_samples, 1)
X2 = 3 * np.random.rand(n_samples, 1)
X = np.hstack((X1, X2))

# Generate target variable (y) with a linear relationship and some noise
y = 4 + 3 * X1 + 5 * X2 + np.random.randn(n_samples, 1) * 2
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.2)
```

```python
print(X.shape, y.shape)
```

## Library Implementation

```python
logreg = LinearRegression()
logreg.fit(X_train, y_train)
y_pred_sklearn = logreg.predict(X_test)
```

```python
sklearn_r2 = r2_score(y_test, y_pred_sklearn)
sklearn_mse = mean_squared_error(y_test, y_pred_sklearn)
print(f'R2 Score with sklearn: {sklearn_r2}')
print(f'MSE Score with sklearn: {sklearn_mse}')
```

## Custom Implementation

```python
class LinearRegressionCustom:
    def __init__(self, learning_rate=0.0001, n_iters=10000):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        # init parameters - ensure weights is a column vector
        self.weights = np.zeros((n_features, 1))
        self.bias = 0
        # gradient descent
        for _ in range(self.n_iters):
            y_predicted = np.dot(X, self.weights) + self.bias

            # compute gradients Ensure y_predicted - y results in a (n_samples, 1) shape
            dw = (1 / n_samples) * np.dot(X.T, (y_predicted - y))
            db = (1 / n_samples) * np.sum(y_predicted - y)

            # update parameters
            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    def predict(self, X):
        y_predicted = np.dot(X, self.weights) + self.bias
        return y_predicted
```

```python
linreg_custom = LinearRegressionCustom(learning_rate=0.0001, n_iters=10000)
linreg_custom.fit(X_train, y_train)
y_pred_custom = linreg_custom.predict(X_test)
```

```python
custom_r2 = r2_score(y_test, y_pred_custom)
custom_mse = mean_squared_error(y_test, y_pred_custom)
print(f"R2 Score with custom implementation: {custom_r2}")
print(f"MSE Score with custom implementation: {custom_mse}")
```
