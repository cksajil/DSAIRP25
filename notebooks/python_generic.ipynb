# Place Holder Passing

```python
a = 2
b = 3.14159
print("a contains {}".format(a))
print("a contains {} and b contains {}".format(a, b))
print("a contains {1} and b contains {0}".format(b, a))
print(f"a contains {a} and b contains {b}")
print(f"b contains {b:.3f}")
```

# List operations

```python
x = [1,2,3]
x.append(4)
print(x)
x.extend([5,6])
print(x)
x.remove(2)
print(x)
x.insert(0,10)
print(x)
x.reverse()
print(x)
x.sort()
print(x)
e = x.pop(0)
print(e, x)
x.clear()
print(x)
```

Write code to convert date format from dd/mm/yyyy to
- mm-dd-yyyy
- dd-mm-yyyy
- yyyy|dd|mm

```python
x = input("Enter the date: ")
dd, mm, yyyy = x.split("/")
print(x)
print("-".join([mm, dd, yyyy]))
print("-".join([dd, mm, yyyy]))
print("|".join([yyyy, dd, mm]))
```
